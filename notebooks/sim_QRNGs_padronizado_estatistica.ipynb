```python
from qiskit import QuantumCircuit, transpile
from qiskit_aer import AerSimulator

import matplotlib.pyplot as plt


# Número de execuções
quantidade_execucoes = 1000


# Cria o circuito
circuito = QuantumCircuit(1, 1)

# Cria superposição
circuito.h(0)

# Mede o qubit
circuito.measure(0, 0)


# Cria o simulador ideal
simulador = AerSimulator()

# Transpila
circuito_compilado = transpile(
    circuito,
    simulador
)

# Executa
job = simulador.run(
    circuito_compilado,
    shots=quantidade_execucoes,
    memory=True
)

resultado = job.result()

bits = resultado.get_memory(
    circuito_compilado
)


# Conta zeros e uns
quantidade_zeros = bits.count("0")
quantidade_uns = bits.count("1")

probabilidade_zero = (
    quantidade_zeros / quantidade_execucoes
)

probabilidade_um = (
    quantidade_uns / quantidade_execucoes
)


print("Quantidade de zeros:", quantidade_zeros)
print("Quantidade de uns:", quantidade_uns)

print(f"P(0): {probabilidade_zero:.2%}")
print(f"P(1): {probabilidade_um:.2%}")


# Gráfico
plt.figure(figsize=(7, 5))

plt.bar(
    ["0", "1"],
    [probabilidade_zero, probabilidade_um]
)

plt.axhline(
    0.5,
    linestyle="--",
    label="Probabilidade ideal: 50%"
)

plt.title("QRNG de 1 qubit — Simulador ideal")
plt.xlabel("Bit")
plt.ylabel("Probabilidade")
plt.ylim(0, 1)

plt.legend()
plt.show()
```

```python
from qiskit import QuantumCircuit, transpile
from qiskit_aer import AerSimulator

import matplotlib.pyplot as plt


numero_qubits = 3
quantidade_execucoes = 1000


circuito = QuantumCircuit(
    numero_qubits,
    numero_qubits
)


# Hadamard em todos os qubits
for qubit in range(numero_qubits):
    circuito.h(qubit)


# Mede todos os qubits
for qubit in range(numero_qubits):
    circuito.measure(
        qubit,
        qubit
    )


simulador = AerSimulator()

circuito_compilado = transpile(
    circuito,
    simulador
)


job = simulador.run(
    circuito_compilado,
    shots=quantidade_execucoes,
    memory=True
)

resultado = job.result()

sequencias = resultado.get_memory(
    circuito_compilado
)


estados = [
    format(numero, f"0{numero_qubits}b")
    for numero in range(2 ** numero_qubits)
]


probabilidades = []

for estado in estados:

    quantidade = sequencias.count(estado)

    probabilidade = (
        quantidade / quantidade_execucoes
    )

    probabilidades.append(probabilidade)

    print(
        estado,
        f"{probabilidade:.2%}"
    )


plt.figure(figsize=(10, 5))

plt.bar(
    estados,
    probabilidades
)

plt.axhline(
    1 / (2 ** numero_qubits),
    linestyle="--",
    label="Probabilidade ideal"
)

plt.title(
    f"QRNG de {numero_qubits} qubits — Simulador ideal"
)

plt.xlabel("Estado")
plt.ylabel("Probabilidade")

plt.legend()
plt.show()
```

```python
from qiskit import QuantumCircuit, transpile
from qiskit_aer import AerSimulator

import math
import matplotlib.pyplot as plt


quantidade_execucoes = 1000

simulador = AerSimulator()


angulos = [
    0,
    math.pi / 6,
    math.pi / 4,
    math.pi / 2,
    3 * math.pi / 4,
    5 * math.pi / 6,
    math.pi
]

nomes = [
    "0",
    "π/6",
    "π/4",
    "π/2",
    "3π/4",
    "5π/6",
    "π"
]


probabilidades_experimentais = []
probabilidades_teoricas = []


for angulo in angulos:

    circuito = QuantumCircuit(1, 1)

    circuito.ry(
        angulo,
        0
    )

    circuito.measure(0, 0)


    circuito_compilado = transpile(
        circuito,
        simulador
    )


    job = simulador.run(
        circuito_compilado,
        shots=quantidade_execucoes,
        memory=True
    )

    resultado = job.result()

    bits = resultado.get_memory(
        circuito_compilado
    )


    probabilidade_um = (
        bits.count("1")
        / quantidade_execucoes
    )

    probabilidade_teorica = (
        math.sin(angulo / 2) ** 2
    )


    probabilidades_experimentais.append(
        probabilidade_um
    )

    probabilidades_teoricas.append(
        probabilidade_teorica
    )


plt.figure(figsize=(10, 5))

plt.plot(
    nomes,
    probabilidades_teoricas,
    marker="o",
    label="Teórica"
)

plt.plot(
    nomes,
    probabilidades_experimentais,
    marker="o",
    label="Simulada"
)

plt.title(
    "QRNG com rotação Ry — Simulador ideal"
)

plt.xlabel("Ângulo")
plt.ylabel("P(1)")

plt.ylim(0, 1)

plt.legend()
plt.show()
```

```python
from qiskit import QuantumCircuit, transpile
from qiskit_aer import AerSimulator
from qiskit_aer.noise import NoiseModel, depolarizing_error

import matplotlib.pyplot as plt


numero_qubits = 8
quantidade_execucoes = 10000


# Cria o circuito
circuito = QuantumCircuit(
    numero_qubits,
    numero_qubits
)

for qubit in range(numero_qubits):
    circuito.h(qubit)

for qubit in range(numero_qubits):
    circuito.measure(qubit, qubit)


# Modelo de ruído
modelo_ruido = NoiseModel()

# Erro de 2% nas portas de um qubit
erro_1_qubit = depolarizing_error(
    0.02,
    1
)

modelo_ruido.add_all_qubit_quantum_error(
    erro_1_qubit,
    ["h"]
)


# Simulador com ruído
simulador = AerSimulator(
    noise_model=modelo_ruido
)


circuito_compilado = transpile(
    circuito,
    simulador
)


job = simulador.run(
    circuito_compilado,
    shots=quantidade_execucoes,
    memory=True
)


resultado = job.result()

sequencias = resultado.get_memory(
    circuito_compilado
)


# Junta todos os bits
bits_ruidosos = "".join(sequencias)


quantidade_zeros = bits_ruidosos.count("0")
quantidade_uns = bits_ruidosos.count("1")

total = len(bits_ruidosos)

p_zero = quantidade_zeros / total
p_um = quantidade_uns / total


print("Quantidade de bits:", total)
print("Zeros:", quantidade_zeros)
print("Uns:", quantidade_uns)

print(f"P(0): {p_zero:.4%}")
print(f"P(1): {p_um:.4%}")


plt.figure(figsize=(7, 5))

plt.bar(
    ["0", "1"],
    [p_zero, p_um]
)

plt.axhline(
    0.5,
    linestyle="--",
    label="Ideal: 50%"
)

plt.title(
    "QRNG em simulador quântico com ruído"
)

plt.xlabel("Bit")
plt.ylabel("Probabilidade")
plt.ylim(0, 1)

plt.legend()
plt.show()
```

```python
from qiskit import QuantumCircuit, transpile
from qiskit_aer import AerSimulator


def converter_alfanumerico(bits):

    caracteres = (
        "0123456789"
        "ABCDEFGHIJKLMNOPQRSTUVWXYZ"
        "abcdefghijklmnopqrstuvwxyz"
    )

    resultado = []

    for indice in range(
        0,
        len(bits) - 5,
        6
    ):

        bloco = bits[
            indice:indice + 6
        ]

        valor = int(
            bloco,
            2
        )

        # 62 e 63 são rejeitados
        if valor < 62:

            resultado.append(
                caracteres[valor]
            )

    return "".join(resultado)


numero_qubits = 8
quantidade_execucoes = 10000


circuito = QuantumCircuit(
    numero_qubits,
    numero_qubits
)


for qubit in range(numero_qubits):
    circuito.h(qubit)

for qubit in range(numero_qubits):
    circuito.measure(
        qubit,
        qubit
    )


simulador = AerSimulator()

circuito_compilado = transpile(
    circuito,
    simulador
)


job = simulador.run(
    circuito_compilado,
    shots=quantidade_execucoes,
    memory=True
)


resultado = job.result()

sequencias = resultado.get_memory(
    circuito_compilado
)


bits_brutos = "".join(
    sequencias
)


sequencia_alfanumerica = (
    converter_alfanumerico(
        bits_brutos
    )
)


print(
    "Quantidade de bits:",
    len(bits_brutos)
)

print(
    "Quantidade de caracteres:",
    len(sequencia_alfanumerica)
)

print(
    "\nPrimeiros 500 caracteres:"
)

print(
    sequencia_alfanumerica[:500]
)
```

```python
from qiskit import QuantumCircuit, transpile
from qiskit_aer import AerSimulator

import string


def converter_caracteres(bits):

    caracteres = (
        string.digits
        + string.ascii_uppercase
        + string.ascii_lowercase
        + string.punctuation
    )

    resultado = []

    # Precisamos de 7 bits
    for indice in range(
        0,
        len(bits) - 6,
        7
    ):

        bloco = bits[
            indice:indice + 7
        ]

        valor = int(
            bloco,
            2
        )

        # Valores acima de 93 são rejeitados
        if valor < len(caracteres):

            resultado.append(
                caracteres[valor]
            )

    return "".join(resultado)


numero_qubits = 8
quantidade_execucoes = 10000


circuito = QuantumCircuit(
    numero_qubits,
    numero_qubits
)


for qubit in range(numero_qubits):
    circuito.h(qubit)

for qubit in range(numero_qubits):
    circuito.measure(
        qubit,
        qubit
    )


simulador = AerSimulator()

circuito_compilado = transpile(
    circuito,
    simulador
)


job = simulador.run(
    circuito_compilado,
    shots=quantidade_execucoes,
    memory=True
)


resultado = job.result()

sequencias = resultado.get_memory(
    circuito_compilado
)


bits_brutos = "".join(
    sequencias
)


sequencia_caracteres = (
    converter_caracteres(
        bits_brutos
    )
)


print(
    "Quantidade de caracteres:",
    len(sequencia_caracteres)
)

print(
    "\nPrimeiros 500:"
)

print(
    sequencia_caracteres[:500]
)
```

```python
from qiskit import QuantumCircuit, transpile
from qiskit_aer import AerSimulator

import matplotlib.pyplot as plt


def extrator_von_neumann(bits):

    resultado = []

    for indice in range(
        0,
        len(bits) - 1,
        2
    ):

        par = bits[
            indice:indice + 2
        ]

        if par == "01":
            resultado.append("0")

        elif par == "10":
            resultado.append("1")

    return "".join(resultado)


numero_qubits = 8
quantidade_execucoes = 10000


circuito = QuantumCircuit(
    numero_qubits,
    numero_qubits
)


for qubit in range(numero_qubits):
    circuito.h(qubit)

for qubit in range(numero_qubits):
    circuito.measure(
        qubit,
        qubit
    )


simulador = AerSimulator()

circuito_compilado = transpile(
    circuito,
    simulador
)


job = simulador.run(
    circuito_compilado,
    shots=quantidade_execucoes,
    memory=True
)


resultado = job.result()

sequencias = resultado.get_memory(
    circuito_compilado
)


bits_brutos = "".join(
    sequencias
)


bits_von_neumann = (
    extrator_von_neumann(
        bits_brutos
    )
)


print(
    "Bits brutos:",
    len(bits_brutos)
)

print(
    "Bits após Von Neumann:",
    len(bits_von_neumann)
)


p0_bruto = (
    bits_brutos.count("0")
    / len(bits_brutos)
)

p1_bruto = (
    bits_brutos.count("1")
    / len(bits_brutos)
)


p0_vn = (
    bits_von_neumann.count("0")
    / len(bits_von_neumann)
)

p1_vn = (
    bits_von_neumann.count("1")
    / len(bits_von_neumann)
)


print("\nAntes:")
print(f"P(0): {p0_bruto:.4%}")
print(f"P(1): {p1_bruto:.4%}")

print("\nApós Von Neumann:")
print(f"P(0): {p0_vn:.4%}")
print(f"P(1): {p1_vn:.4%}")


plt.figure(figsize=(9, 5))

plt.bar(
    [
        "Bruto 0",
        "Bruto 1",
        "Von Neumann 0",
        "Von Neumann 1"
    ],
    [
        p0_bruto,
        p1_bruto,
        p0_vn,
        p1_vn
    ]
)

plt.axhline(
    0.5,
    linestyle="--",
    label="Ideal: 50%"
)

plt.title(
    "QRNG simulado antes e depois de Von Neumann"
)

plt.ylabel("Proporção")
plt.ylim(0, 1)

plt.legend()
plt.show()
```

```python
from qiskit import QuantumCircuit, transpile
from qiskit_aer import AerSimulator

import hashlib
import matplotlib.pyplot as plt


def extrator_sha256(
    bits,
    tamanho_bloco=1024
):

    saida = []

    quantidade_blocos = (
        len(bits)
        // tamanho_bloco
    )

    for numero_bloco in range(
        quantidade_blocos
    ):

        inicio = (
            numero_bloco
            * tamanho_bloco
        )

        fim = (
            inicio
            + tamanho_bloco
        )

        bloco = bits[
            inicio:fim
        ]

        bloco_bytes = int(
            bloco,
            2
        ).to_bytes(
            tamanho_bloco // 8,
            byteorder="big"
        )


        hash_bytes = hashlib.sha256(
            bloco_bytes
        ).digest()


        hash_bits = "".join(
            format(byte, "08b")
            for byte in hash_bytes
        )


        saida.append(
            hash_bits
        )

    return "".join(saida)


numero_qubits = 8
quantidade_execucoes = 10000


circuito = QuantumCircuit(
    numero_qubits,
    numero_qubits
)


for qubit in range(numero_qubits):
    circuito.h(qubit)

for qubit in range(numero_qubits):
    circuito.measure(
        qubit,
        qubit
    )


simulador = AerSimulator()

circuito_compilado = transpile(
    circuito,
    simulador
)


job = simulador.run(
    circuito_compilado,
    shots=quantidade_execucoes,
    memory=True
)


resultado = job.result()

sequencias = resultado.get_memory(
    circuito_compilado
)


bits_brutos = "".join(
    sequencias
)


bits_sha256 = extrator_sha256(
    bits_brutos
)


p0_bruto = (
    bits_brutos.count("0")
    / len(bits_brutos)
)

p1_bruto = (
    bits_brutos.count("1")
    / len(bits_brutos)
)


p0_sha = (
    bits_sha256.count("0")
    / len(bits_sha256)
)

p1_sha = (
    bits_sha256.count("1")
    / len(bits_sha256)
)


print(
    "Bits brutos:",
    len(bits_brutos)
)

print(
    "Bits após SHA-256:",
    len(bits_sha256)
)


print("\nSequência bruta:")
print(f"P(0): {p0_bruto:.4%}")
print(f"P(1): {p1_bruto:.4%}")


print("\nApós SHA-256:")
print(f"P(0): {p0_sha:.4%}")
print(f"P(1): {p1_sha:.4%}")


plt.figure(figsize=(9, 5))

plt.bar(
    [
        "Bruto 0",
        "Bruto 1",
        "SHA-256 0",
        "SHA-256 1"
    ],
    [
        p0_bruto,
        p1_bruto,
        p0_sha,
        p1_sha
    ]
)

plt.axhline(
    0.5,
    linestyle="--",
    label="Ideal: 50%"
)

plt.title(
    "QRNG simulado antes e depois do SHA-256"
)

plt.ylabel("Proporção")
plt.ylim(0, 1)

plt.legend()
plt.show()
```

Controle ideal: 1 qubit no estado |0⟩, porta Hadamard e medição. São gerados 30 lotes × 10.000 bits.


```python

import math
import os
from pathlib import Path

import numpy as np
import pandas as pd
import statsmodels
import scipy


NUM_LOTES = 30
BITS_POR_LOTE = 10_000
ALFA = 0.05
Z_95 = 1.959963984540054

print("Python/ambiente estatístico:")
print("statsmodels:", statsmodels.__version__)
print("scipy:", scipy.__version__)


def intervalo_wilson(p, n, z=Z_95):
    if n == 0:
        return np.nan, np.nan

    denominador = 1 + (z ** 2) / n
    centro = (p + (z ** 2) / (2 * n)) / denominador
    margem = (
        z
        * math.sqrt(
            (p * (1 - p) + (z ** 2) / (4 * n))
            / n
        )
        / denominador
    )

    return centro - margem, centro + margem


def runs_test(bits):
    """
    Runs Test (Wald-Wolfowitz) usando statsmodels.

    H0: a ordem dos bits é compatível com aleatoriedade em relação
        ao número de corridas.
    H1: o número de corridas difere do esperado sob H0.

    Regra de decisão:
        p >= ALFA -> não rejeita H0
        p <  ALFA -> rejeita H0
    """
    from statsmodels.sandbox.stats.runs import runstest_1samp

    vetor = np.fromiter(
        (1 if bit == "1" else 0 for bit in bits),
        dtype=np.int8
    )

    if len(vetor) < 2 or np.all(vetor == vetor[0]):
        return np.nan, np.nan, np.nan

    numero_runs = 1 + int(np.sum(vetor[1:] != vetor[:-1]))

    z, p_valor = runstest_1samp(
        vetor,
        cutoff=0.5,
        correction=False
    )

    return numero_runs, float(z), float(p_valor)



def calcular_metricas(bits, fonte, lote, condicao="bruta"):
    bits = "".join(bits)

    n = len(bits)

    if n == 0:
        raise ValueError("O lote não contém bits.")

    zeros = bits.count("0")
    uns = bits.count("1")

    p0 = zeros / n
    p1 = uns / n

    vies = abs(p1 - 0.5)

    shannon = 0.0

    for p in (p0, p1):
        if p > 0:
            shannon -= p * math.log2(p)

    min_entropia = -math.log2(
        max(p0, p1)
    )

    vetor = np.fromiter(
        (1 if bit == "1" else 0 for bit in bits),
        dtype=np.float64
    )

    if (
        n > 1
        and np.std(vetor[:-1]) > 0
        and np.std(vetor[1:]) > 0
    ):
        autocorrelacao_lag1 = float(
            np.corrcoef(
                vetor[:-1],
                vetor[1:]
            )[0, 1]
        )
    else:
        autocorrelacao_lag1 = np.nan

    numero_runs, runs_z, runs_p_valor = runs_test(bits)

    ic95_inferior, ic95_superior = intervalo_wilson(
        p1,
        n
    )

    return {
        "fonte": fonte,
        "condicao": condicao,
        "lote": lote,
        "n_bits": n,
        "zeros": zeros,
        "uns": uns,
        "p0": p0,
        "p1": p1,
        "vies": vies,
        "ic95_p1_inferior": ic95_inferior,
        "ic95_p1_superior": ic95_superior,
        "entropia_shannon": shannon,
        "min_entropia_empirica": min_entropia,
        "autocorrelacao_lag1": autocorrelacao_lag1,
        "numero_runs": numero_runs,
        "runs_z": runs_z,
        "runs_p_valor": runs_p_valor,
        "runs_aprovado_alfa_005": (
            runs_p_valor >= ALFA
            if not np.isnan(runs_p_valor)
            else False
        ),
    }


def salvar_lote(bits, pasta, nome):
    pasta = Path(pasta)
    pasta.mkdir(
        parents=True,
        exist_ok=True
    )

    caminho = pasta / nome

    caminho.write_text(
        "".join(bits),
        encoding="utf-8"
    )

    return caminho


def resumo_metricas(df):
    colunas = [
        "p1",
        "vies",
        "entropia_shannon",
        "min_entropia_empirica",
        "autocorrelacao_lag1",
        "runs_z",
        "runs_p_valor",
    ]

    print("\nResumo dos 30 lotes:")
    display(
        df[colunas]
        .agg(["mean", "std", "min", "median", "max"])
        .T
    )

    print(
        "\nProporção de aprovação no Runs Test (alfa = 0,05):",
        f"{df['runs_aprovado_alfa_005'].mean():.2%}"
    )


# EXPERIMENTO PADRONIZADO — SIMULADOR IDEAL QISKIT

from qiskit import QuantumCircuit, transpile
from qiskit_aer import AerSimulator

FONTE = "Qiskit_AerSimulator_ideal"
SEMENTE_SIMULADOR_BASE = 1000

pasta_saida = Path(
    "/content/dados_padronizados/simulador"
)

# Mesmo circuito lógico usado na comparação com a QPU:
# |0> -> H -> medição

circuito = QuantumCircuit(1, 1)
circuito.h(0)
circuito.measure(0, 0)

simulador = AerSimulator()

circuito_compilado = transpile(
    circuito,
    simulador,
    optimization_level=0,
    seed_transpiler=7
)

resultados = []

for lote in range(1, NUM_LOTES + 1):

    # Sementes diferentes por lote tornam o controle
    # reprodutível sem repetir a mesma amostra.
    semente_lote = (
        SEMENTE_SIMULADOR_BASE
        + lote
    )

    job = simulador.run(
        circuito_compilado,
        shots=BITS_POR_LOTE,
        memory=True,
        seed_simulator=semente_lote
    )

    resultado = job.result()

    memoria = resultado.get_memory(
        circuito_compilado
    )

    bits = "".join(memoria)

    salvar_lote(
        bits,
        pasta_saida,
        f"simulador_lote_{lote:02d}.txt"
    )

    metricas = calcular_metricas(
        bits,
        fonte=FONTE,
        lote=lote
    )

    metricas["seed_simulator"] = (
        semente_lote
    )

    resultados.append(metricas)

df_simulador = pd.DataFrame(resultados)

caminho_csv = (
    pasta_saida
    / "simulador_metricas_30x10000.csv"
)

df_simulador.to_csv(
    caminho_csv,
    index=False
)

print("Experimento concluído.")
print("Circuito: 1 qubit |0> -> H -> medição")
print("Lotes:", NUM_LOTES)
print("Shots/bits por lote:", BITS_POR_LOTE)
print("Total de bits:", NUM_LOTES * BITS_POR_LOTE)
print("CSV:", caminho_csv)

display(df_simulador.head())

resumo_metricas(df_simulador)
```
