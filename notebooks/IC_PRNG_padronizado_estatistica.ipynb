```python
import random
import math
import matplotlib.pyplot as plt


# Quantidade de bits
QUANTIDADE_BITS = 10000

# Semente para reproduzir o experimento
SEMENTE = 42



# GERAÇÃO DOS BITS PSEUDOALEATÓRIOS


random.seed(SEMENTE)

bits = [
    random.randint(0, 1)
    for _ in range(QUANTIDADE_BITS)
]



# CÁLCULO DAS PROBABILIDADES


total = len(bits)

quantidade_zeros = bits.count(0)
quantidade_uns = bits.count(1)

p_zero = quantidade_zeros / total
p_um = quantidade_uns / total



# ENTROPIA DE SHANNON


entropia = 0

if p_zero > 0:
    entropia -= p_zero * math.log2(p_zero)

if p_um > 0:
    entropia -= p_um * math.log2(p_um)



# RESULTADOS


print("Primeiros 100 bits:")
print("".join(map(str, bits[:100])))

print("\nResultados:")
print(f"Total de bits: {total}")
print(f"P(0): {p_zero:.6f}")
print(f"P(1): {p_um:.6f}")
print(f"Entropia de Shannon: {entropia:.6f} bits")
print("Entropia máxima: 1.000000 bit")



# GRÁFICO


categorias = [
    "Entropia obtida",
    "Entropia máxima"
]

valores = [
    entropia,
    1.0
]

plt.figure(figsize=(7, 5))

barras = plt.bar(
    categorias,
    valores
)

plt.title("Entropia de Shannon — PRNG")
plt.ylabel("Entropia em bits")
plt.ylim(0, 1.1)

for barra, valor in zip(barras, valores):
    plt.text(
        barra.get_x() + barra.get_width() / 2,
        valor + 0.02,
        f"{valor:.6f}",
        ha="center"
    )

plt.tight_layout()
plt.show()
```

```python
import random
import numpy as np
import matplotlib.pyplot as plt


# Quantidade de bits
QUANTIDADE_BITS = 10000

# Semente para reproduzir o experimento
SEMENTE = 42

# Maior atraso analisado
ATRASO_MAXIMO = 50



# GERAÇÃO DOS BITS PSEUDOALEATÓRIOS


random.seed(SEMENTE)

bits = [
    random.randint(0, 1)
    for _ in range(QUANTIDADE_BITS)
]



# FUNÇÃO DE AUTOCORRELAÇÃO


def calcular_autocorrelacao(sequencia, atraso):
    """
    Calcula a correlação entre a sequência original
    e uma versão deslocada dela.
    """

    if atraso <= 0:
        raise ValueError("O atraso deve ser maior que zero.")

    if atraso >= len(sequencia):
        raise ValueError(
            "O atraso deve ser menor que o tamanho da sequência."
        )

    valores = np.array(sequencia, dtype=float)

    primeira_parte = valores[:-atraso]
    segunda_parte = valores[atraso:]

    desvio_primeira = np.std(primeira_parte)
    desvio_segunda = np.std(segunda_parte)

    if desvio_primeira == 0 or desvio_segunda == 0:
        return 0.0

    correlacao = np.corrcoef(
        primeira_parte,
        segunda_parte
    )[0, 1]

    return float(correlacao)



# CÁLCULO PARA VÁRIOS ATRASOS


atrasos = list(range(1, ATRASO_MAXIMO + 1))

autocorrelacoes = [
    calcular_autocorrelacao(bits, atraso)
    for atraso in atrasos
]



# RESULTADOS


print("Primeiros 100 bits:")
print("".join(map(str, bits[:100])))

print("\nAutocorrelações:")

for atraso, correlacao in zip(
    atrasos,
    autocorrelacoes
):
    print(
        f"Atraso {atraso:2d}: "
        f"{correlacao:.6f}"
    )

print("\nAutocorrelação com atraso 1:")
print(f"{autocorrelacoes[0]:.6f}")

print("\nValor ideal:")
print("Próximo de 0")



# GRÁFICO


plt.figure(figsize=(11, 5))

plt.stem(
    atrasos,
    autocorrelacoes
)

plt.axhline(
    y=0,
    linestyle="--",
    label="Autocorrelação ideal"
)

plt.title("Autocorrelação da sequência — PRNG")
plt.xlabel("Atraso")
plt.ylabel("Coeficiente de autocorrelação")

plt.grid(alpha=0.3)
plt.legend()

plt.tight_layout()
plt.show()
```

```python
import random
import math
import matplotlib.pyplot as plt


# Quantidade de bits
QUANTIDADE_BITS = 10000

# Semente para reproduzir o experimento
SEMENTE = 42


# GERAÇÃO DOS BITS PSEUDOALEATÓRIOS

random.seed(SEMENTE)

bits = [
    random.randint(0, 1)
    for _ in range(QUANTIDADE_BITS)
]


# RUNS TEST

total = len(bits)

quantidade_zeros = bits.count(0)
quantidade_uns = bits.count(1)

# Conta quantas sequências consecutivas existem
quantidade_runs = 1

for indice in range(1, total):
    if bits[indice] != bits[indice - 1]:
        quantidade_runs += 1


# Quantidade esperada de runs
runs_esperadas = (
    1
    + (2 * quantidade_zeros * quantidade_uns) / total
)


# Variância da quantidade de runs
variancia = (
    2
    * quantidade_zeros
    * quantidade_uns
    * (
        2
        * quantidade_zeros
        * quantidade_uns
        - total
    )
) / (
    total ** 2
    * (total - 1)
)


desvio_padrao = math.sqrt(variancia)


# Z-score
z_score = (
    quantidade_runs - runs_esperadas
) / desvio_padrao


# P-valor bilateral
p_valor = math.erfc(
    abs(z_score) / math.sqrt(2)
)


# RESULTADOS

print("Primeiros 100 bits:")
print("".join(map(str, bits[:100])))

print("\nResultados do Runs Test:")
print(f"Total de bits: {total}")
print(f"Quantidade de zeros: {quantidade_zeros}")
print(f"Quantidade de uns: {quantidade_uns}")
print(f"Runs observadas: {quantidade_runs}")
print(f"Runs esperadas: {runs_esperadas:.6f}")
print(f"Variância: {variancia:.6f}")
print(f"Desvio-padrão: {desvio_padrao:.6f}")
print(f"Z-score: {z_score:.6f}")
print(f"P-valor: {p_valor:.6f}")


# Interpretação com nível de significância de 5%
if p_valor >= 0.05:
    print("\nResultado: sequência aprovada no runs test.")
else:
    print("\nResultado: sequência reprovada no runs test.")


# GRÁFICO

categorias = [
    "Runs observadas",
    "Runs esperadas"
]

valores = [
    quantidade_runs,
    runs_esperadas
]

plt.figure(figsize=(8, 5))

barras = plt.bar(
    categorias,
    valores
)

plt.title(
    f"Runs Test — PRNG\n"
    f"p-valor = {p_valor:.6f}"
)

plt.ylabel("Quantidade de runs")

for barra, valor in zip(barras, valores):
    plt.text(
        barra.get_x() + barra.get_width() / 2,
        valor,
        f"{valor:.2f}",
        ha="center",
        va="bottom"
    )

plt.tight_layout()
plt.show()
```

```python
import random
import zlib
import matplotlib.pyplot as plt


# Quantidade de bits
QUANTIDADE_BITS = 1_000_000

# Semente para reproduzir o experimento
SEMENTE = 42

# GERAÇÃO DOS BITS PSEUDOALEATÓRIOS
random.seed(SEMENTE)

bits = [
    random.randint(0, 1)
    for _ in range(QUANTIDADE_BITS)
]

# CONVERSÃO DOS BITS PARA BYTES
bytes_gerados = bytearray()

for i in range(0, len(bits) - 7, 8):

    grupo = bits[i:i + 8]

    binario = "".join(
        str(bit)
        for bit in grupo
    )

    valor = int(binario, 2)

    bytes_gerados.append(valor)


# Converte para bytes
dados_originais = bytes(bytes_gerados)



# COMPRESSÃO


dados_comprimidos = zlib.compress(
    dados_originais,
    level=9
)



# TAMANHOS


tamanho_original = len(dados_originais)

tamanho_comprimido = len(dados_comprimidos)



# TAXA DE COMPRESSÃO


taxa_compressao = (
    tamanho_comprimido
    / tamanho_original
)



# REDUÇÃO PERCENTUAL


reducao_percentual = (
    1 - taxa_compressao
) * 100



# RESULTADOS


print("Quantidade de bits:")
print(len(bits))

print("\nQuantidade de bytes:")
print(len(bytes_gerados))

print("\nTamanho original:")
print(f"{tamanho_original} bytes")

print("\nTamanho comprimido:")
print(f"{tamanho_comprimido} bytes")

print("\nTaxa de compressão:")
print(f"{taxa_compressao:.6f}")

print("\nRedução percentual:")
print(f"{reducao_percentual:.4f}%")



# INTERPRETAÇÃO


print("\nInterpretação:")

if tamanho_comprimido >= tamanho_original:

    print(
        "A sequência não foi comprimida."
    )

    print(
        "O compressor produziu um arquivo "
        "igual ou maior que o original."
    )

else:

    print(
        "A sequência apresentou alguma compressibilidade."
    )



# GRÁFICO


categorias = [
    "Original",
    "Comprimido"
]

tamanhos = [
    tamanho_original,
    tamanho_comprimido
]

plt.figure(figsize=(8, 5))

barras = plt.bar(
    categorias,
    tamanhos
)

plt.title("Compressibilidade — PRNG")

plt.ylabel("Tamanho em bytes")


for barra, valor in zip(barras, tamanhos):

    plt.text(
        barra.get_x() + barra.get_width() / 2,
        valor,
        f"{valor} bytes",
        ha="center",
        va="bottom"
    )


plt.tight_layout()

plt.show()
```

```python
import random
import matplotlib.pyplot as plt
from scipy.stats import chi2


# Quantidade de bits
QUANTIDADE_BITS = 1_000_000

# Semente para reproduzir o experimento
SEMENTE = 42

# Nível de significância
ALFA = 0.05


# GERAÇÃO DOS BITS PSEUDOALEATÓRIOS

random.seed(SEMENTE)

bits = [
    random.randint(0, 1)
    for _ in range(QUANTIDADE_BITS)
]


# CONVERSÃO DOS BITS PARA BYTES

bytes_gerados = []

for i in range(0, len(bits) - 7, 8):
    grupo = bits[i:i + 8]

    texto_binario = "".join(
        str(bit)
        for bit in grupo
    )

    valor = int(texto_binario, 2)

    bytes_gerados.append(valor)


# CONTAGEM DAS FREQUÊNCIAS

frequencias_observadas = [0] * 256

for valor in bytes_gerados:
    frequencias_observadas[valor] += 1


total_bytes = len(bytes_gerados)

frequencia_esperada = total_bytes / 256


# ESTATÍSTICA QUI-QUADRADO

qui_quadrado = 0

for observado in frequencias_observadas:
    qui_quadrado += (
        (observado - frequencia_esperada) ** 2
        / frequencia_esperada
    )


# Número de categorias - 1
graus_liberdade = 256 - 1


# P-valor
p_valor = chi2.sf(
    qui_quadrado,
    graus_liberdade
)


# Valor crítico
valor_critico = chi2.ppf(
    1 - ALFA,
    graus_liberdade
)


# RESULTADOS

print("TESTE QUI-QUADRADO DE UNIFORMIDADE")

print(f"\nQuantidade de bits: {len(bits)}")
print(f"Quantidade de bytes: {total_bytes}")

print(
    f"Frequência esperada por valor: "
    f"{frequencia_esperada:.6f}"
)

print(
    f"Estatística qui-quadrado: "
    f"{qui_quadrado:.6f}"
)

print(
    f"Graus de liberdade: "
    f"{graus_liberdade}"
)

print(
    f"Valor crítico para alfa = {ALFA}: "
    f"{valor_critico:.6f}"
)

print(
    f"P-valor: "
    f"{p_valor:.6f}"
)


# INTERPRETAÇÃO

if p_valor >= ALFA:
    print(
        "\nResultado: não rejeitamos a hipótese "
        "de distribuição uniforme."
    )
else:
    print(
        "\nResultado: rejeitamos a hipótese "
        "de distribuição uniforme."
    )


# GRÁFICO DA DISTRIBUIÇÃO

valores = list(range(256))

plt.figure(figsize=(14, 6))

plt.bar(
    valores,
    frequencias_observadas,
    width=1
)

plt.axhline(
    y=frequencia_esperada,
    linestyle="--",
    label=f"Frequência esperada = {frequencia_esperada:.2f}"
)

plt.title(
    "Teste Qui-Quadrado de Uniformidade — PRNG"
)

plt.xlabel("Valor do byte")
plt.ylabel("Frequência observada")

plt.xlim(0, 255)

plt.legend()
plt.grid(axis="y", alpha=0.3)

plt.tight_layout()
plt.show()


# GRÁFICO DO RESULTADO DO TESTE

categorias = [
    "Qui-quadrado observado",
    "Valor crítico"
]

valores_teste = [
    qui_quadrado,
    valor_critico
]

plt.figure(figsize=(8, 5))

barras = plt.bar(
    categorias,
    valores_teste
)

plt.title(
    f"Resultado do Qui-Quadrado\n"
    f"p-valor = {p_valor:.6f}"
)

plt.ylabel("Valor da estatística")

for barra, valor in zip(
    barras,
    valores_teste
):
    plt.text(
        barra.get_x() + barra.get_width() / 2,
        valor,
        f"{valor:.2f}",
        ha="center",
        va="bottom"
    )

plt.tight_layout()
plt.show()
```

```python
#Teste Serial de Pares de Bits

import random
import matplotlib.pyplot as plt
from scipy.stats import chi2


# Quantidade total de bits
QUANTIDADE_BITS = 1_000_000

# Semente para reproduzir o experimento
SEMENTE = 42

# Nível de significância
ALFA = 0.05



# GERAÇÃO DOS BITS PSEUDOALEATÓRIOS


random.seed(SEMENTE)

bits = [
    random.randint(0, 1)
    for _ in range(QUANTIDADE_BITS)
]



# CONTAGEM DOS PARES SOBREPOSTOS


frequencias = {
    "00": 0,
    "01": 0,
    "10": 0,
    "11": 0
}

for indice in range(len(bits) - 1):

    par = (
        str(bits[indice])
        + str(bits[indice + 1])
    )

    frequencias[par] += 1


# Número total de pares analisados
total_pares = len(bits) - 1

# Frequência esperada para cada par
frequencia_esperada = total_pares / 4



# PROPORÇÕES DOS PARES


proporcoes = {
    par: frequencia / total_pares
    for par, frequencia in frequencias.items()
}



# ESTATÍSTICA QUI-QUADRADO


qui_quadrado = sum(
    (
        frequencias[par]
        - frequencia_esperada
    ) ** 2
    / frequencia_esperada

    for par in frequencias
)


# Quatro categorias menos uma
graus_liberdade = 3


# P-valor
p_valor = chi2.sf(
    qui_quadrado,
    graus_liberdade
)


# Valor crítico
valor_critico = chi2.ppf(
    1 - ALFA,
    graus_liberdade
)



# RESULTADOS


print("TESTE SERIAL DE PARES DE BITS")

print(
    f"\nQuantidade de bits: "
    f"{len(bits)}"
)

print(
    f"Quantidade de pares: "
    f"{total_pares}"
)

print(
    f"Frequência esperada por par: "
    f"{frequencia_esperada:.2f}"
)

print("\nFrequências observadas:")

for par in ["00", "01", "10", "11"]:

    print(
        f"Par {par}: "
        f"{frequencias[par]} "
        f"({proporcoes[par]:.6f})"
    )

print(
    f"\nQui-quadrado observado: "
    f"{qui_quadrado:.6f}"
)

print(
    f"Graus de liberdade: "
    f"{graus_liberdade}"
)

print(
    f"Valor crítico: "
    f"{valor_critico:.6f}"
)

print(
    f"P-valor: "
    f"{p_valor:.6f}"
)



# INTERPRETAÇÃO


if p_valor >= ALFA:

    print(
        "\nResultado: não foi detectado um desvio "
        "significativo na distribuição dos pares."
    )

else:

    print(
        "\nResultado: foi detectado um desvio "
        "significativo na distribuição dos pares."
    )



# GRÁFICO DAS FREQUÊNCIAS


pares = ["00", "01", "10", "11"]

valores_frequencia = [
    frequencias[par]
    for par in pares
]

plt.figure(figsize=(8, 5))

barras = plt.bar(
    pares,
    valores_frequencia
)

plt.axhline(
    y=frequencia_esperada,
    linestyle="--",
    label="Frequência esperada"
)

plt.title(
    "Distribuição dos pares de bits — PRNG"
)

plt.xlabel("Par de bits")
plt.ylabel("Frequência")

plt.legend()

for barra, valor in zip(
    barras,
    valores_frequencia
):
    plt.text(
        barra.get_x()
        + barra.get_width() / 2,
        valor,
        str(valor),
        ha="center",
        va="bottom"
    )

plt.tight_layout()
plt.show()



# GRÁFICO DAS PROPORÇÕES


valores_proporcao = [
    proporcoes[par]
    for par in pares
]

plt.figure(figsize=(8, 5))

barras = plt.bar(
    pares,
    valores_proporcao
)

plt.axhline(
    y=0.25,
    linestyle="--",
    label="Proporção ideal: 0,25"
)

plt.title(
    "Proporção dos pares de bits — PRNG"
)

plt.xlabel("Par de bits")
plt.ylabel("Proporção")

plt.ylim(0, 0.5)
plt.legend()

for barra, valor in zip(
    barras,
    valores_proporcao
):
    plt.text(
        barra.get_x()
        + barra.get_width() / 2,
        valor + 0.01,
        f"{valor:.6f}",
        ha="center"
    )

plt.tight_layout()
plt.show()



# GRÁFICO DO RESULTADO ESTATÍSTICO


categorias = [
    "Qui-quadrado observado",
    "Valor crítico"
]

valores_teste = [
    qui_quadrado,
    valor_critico
]

plt.figure(figsize=(8, 5))

barras = plt.bar(
    categorias,
    valores_teste
)

plt.title(
    f"Teste Serial de Pares\n"
    f"p-valor = {p_valor:.6f}"
)

plt.ylabel("Valor da estatística")

for barra, valor in zip(
    barras,
    valores_teste
):
    plt.text(
        barra.get_x()
        + barra.get_width() / 2,
        valor,
        f"{valor:.4f}",
        ha="center",
        va="bottom"
    )

plt.tight_layout()
plt.show()
```

```python
#Teste da Maior Sequência de Bits 1

import random
import matplotlib.pyplot as plt
from scipy.stats import chi2

# CONFIGURAÇÕES
QUANTIDADE_BITS = 1_000_000
TAMANHO_BLOCO = 128
SEMENTE = 42
ALFA = 0.05



# GERAÇÃO DOS BITS PSEUDOALEATÓRIOS
random.seed(SEMENTE)

bits = [
    random.randint(0, 1)
    for _ in range(QUANTIDADE_BITS)
]



# FUNÇÃO PARA ENCONTRAR A MAIOR SEQUÊNCIA DE 1
def maior_sequencia_de_uns(bloco):
    maior = 0
    atual = 0

    for bit in bloco:
        if bit == 1:
            atual += 1

            if atual > maior:
                maior = atual
        else:
            atual = 0

    return maior



# DIVISÃO EM BLOCOS


quantidade_blocos = len(bits) // TAMANHO_BLOCO

bits_utilizados = bits[
    :quantidade_blocos * TAMANHO_BLOCO
]


maiores_sequencias = []

for indice in range(quantidade_blocos):
    inicio = indice * TAMANHO_BLOCO
    fim = inicio + TAMANHO_BLOCO

    bloco = bits_utilizados[inicio:fim]

    maior_run = maior_sequencia_de_uns(bloco)

    maiores_sequencias.append(maior_run)

# CATEGORIAS DO TESTE PARA BLOCOS DE 128 BITS

# Categorias:
# <= 4
# 5
# 6
# 7
# 8
# >= 9

categorias = [
    "<= 4",
    "5",
    "6",
    "7",
    "8",
    ">= 9"
]


frequencias_observadas = [0] * 6


for valor in maiores_sequencias:

    if valor <= 4:
        frequencias_observadas[0] += 1

    elif valor == 5:
        frequencias_observadas[1] += 1

    elif valor == 6:
        frequencias_observadas[2] += 1

    elif valor == 7:
        frequencias_observadas[3] += 1

    elif valor == 8:
        frequencias_observadas[4] += 1

    else:
        frequencias_observadas[5] += 1



# PROBABILIDADES ESPERADAS


# Valores usados pelo teste NIST para blocos de 128 bits

probabilidades_esperadas = [
    0.1174035788,
    0.2429559590,
    0.2493634830,
    0.1751770600,
    0.1027010710,
    0.1123981520
]


frequencias_esperadas = [
    quantidade_blocos * probabilidade
    for probabilidade in probabilidades_esperadas
]



# ESTATÍSTICA QUI-QUADRADO


qui_quadrado = sum(
    (
        observado - esperado
    ) ** 2 / esperado

    for observado, esperado in zip(
        frequencias_observadas,
        frequencias_esperadas
    )
)


graus_liberdade = len(categorias) - 1


p_valor = chi2.sf(
    qui_quadrado,
    graus_liberdade
)


valor_critico = chi2.ppf(
    1 - ALFA,
    graus_liberdade
)



# RESULTADOS


print(" MAIOR SEQUÊNCIA DE BITS 1")

print(
    f"\nQuantidade total de bits: "
    f"{len(bits)}"
)

print(
    f"Bits utilizados: "
    f"{len(bits_utilizados)}"
)

print(
    f"Tamanho de cada bloco: "
    f"{TAMANHO_BLOCO}"
)

print(
    f"Quantidade de blocos: "
    f"{quantidade_blocos}"
)

print(
    f"Maior sequência encontrada em toda a amostra: "
    f"{max(maiores_sequencias)}"
)


print("\nFrequências por categoria:")

for categoria, observado, esperado in zip(
    categorias,
    frequencias_observadas,
    frequencias_esperadas
):
    print(
        f"{categoria:>4}: "
        f"observado = {observado:5d} | "
        f"esperado = {esperado:.2f}"
    )


print(
    f"\nQui-quadrado observado: "
    f"{qui_quadrado:.6f}"
)

print(
    f"Graus de liberdade: "
    f"{graus_liberdade}"
)

print(
    f"Valor crítico: "
    f"{valor_critico:.6f}"
)

print(
    f"P-valor: "
    f"{p_valor:.6f}"
)



# INTERPRETAÇÃO


if p_valor >= ALFA:

    print(
        "\nResultado: a sequência foi aprovada "
        "no teste da maior sequência de bits 1."
    )

else:

    print(
        "\nResultado: a sequência foi reprovada "
        "no teste da maior sequência de bits 1."
    )



# GRÁFICO DAS FREQUÊNCIAS


x = list(range(len(categorias)))

largura = 0.38

plt.figure(figsize=(10, 6))

plt.bar(
    [valor - largura / 2 for valor in x],
    frequencias_observadas,
    width=largura,
    label="Observado"
)

plt.bar(
    [valor + largura / 2 for valor in x],
    frequencias_esperadas,
    width=largura,
    label="Esperado"
)

plt.xticks(x, categorias)

plt.title(
    "Maior Sequência de Bits 1 por Bloco — PRNG"
)

plt.xlabel("Comprimento da maior sequência")
plt.ylabel("Quantidade de blocos")

plt.legend()
plt.grid(axis="y", alpha=0.3)

plt.tight_layout()
plt.show()



# HISTOGRAMA DAS MAIORES SEQUÊNCIAS


plt.figure(figsize=(10, 5))

plt.hist(
    maiores_sequencias,
    bins=range(
        min(maiores_sequencias),
        max(maiores_sequencias) + 2
    ),
    align="left",
    edgecolor="black"
)

plt.title(
    "Distribuição das Maiores Sequências de Bits 1"
)

plt.xlabel("Comprimento da maior sequência")
plt.ylabel("Quantidade de blocos")

plt.grid(axis="y", alpha=0.3)

plt.tight_layout()
plt.show()



# GRÁFICO DO RESULTADO ESTATÍSTICO


valores_teste = [
    qui_quadrado,
    valor_critico
]

rotulos_teste = [
    "Qui-quadrado observado",
    "Valor crítico"
]

plt.figure(figsize=(8, 5))

barras = plt.bar(
    rotulos_teste,
    valores_teste
)

plt.title(
    f"Teste da Maior Sequência de Bits 1\n"
    f"p-valor = {p_valor:.6f}"
)

plt.ylabel("Valor da estatística")

for barra, valor in zip(
    barras,
    valores_teste
):
    plt.text(
        barra.get_x()
        + barra.get_width() / 2,
        valor,
        f"{valor:.4f}",
        ha="center",
        va="bottom"
    )

plt.tight_layout()
plt.show()
```

```python

# TESTE DE SOMAS ACUMULADAS

import random
import math

import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm


# CONFIGURAÇÕES

QUANTIDADE_BITS = 1_000_000
SEMENTE = 42
ALFA = 0.05


# GERAÇÃO DOS BITS PSEUDOALEATÓRIOS

random.seed(SEMENTE)

bits = [
    random.randint(0, 1)
    for _ in range(QUANTIDADE_BITS)
]


# VALIDAÇÃO DA SEQUÊNCIA

def validar_bits(sequencia):
    if len(sequencia) == 0:
        raise ValueError("A sequência está vazia.")

    if any(bit not in (0, 1) for bit in sequencia):
        raise ValueError(
            "A sequência deve conter apenas bits 0 e 1."
        )


# CONVERSÃO PARA CAMINHADA ALEATÓRIA

def criar_somas_acumuladas(sequencia):
    """
    Converte:

        0 -> -1
        1 -> +1

    Em seguida, calcula as somas acumuladas.
    """

    validar_bits(sequencia)

    valores = np.array(
        [
            1 if bit == 1 else -1
            for bit in sequencia
        ],
        dtype=int
    )

    somas = np.cumsum(valores)

    return valores, somas


# CÁLCULO DO P-VALOR DO TESTE NIST

def calcular_p_valor_somas_acumuladas(
    tamanho_sequencia,
    maior_excursao
):
    """
    Calcula o p-valor do teste Cumulative Sums,
    conforme a formulação do NIST SP 800-22.

    tamanho_sequencia = número de bits
    maior_excursao = maior valor absoluto da soma acumulada
    """

    n = tamanho_sequencia
    z = maior_excursao

    if z == 0:
        return 1.0

    raiz_n = math.sqrt(n)

    soma_1 = 0.0

    limite_inferior_1 = math.floor(
        ((-n / z) + 1) / 4
    )

    limite_superior_1 = math.floor(
        ((n / z) - 1) / 4
    )

    for k in range(
        limite_inferior_1,
        limite_superior_1 + 1
    ):
        termo_superior = norm.cdf(
            ((4 * k + 1) * z) / raiz_n
        )

        termo_inferior = norm.cdf(
            ((4 * k - 1) * z) / raiz_n
        )

        soma_1 += (
            termo_superior
            - termo_inferior
        )

    soma_2 = 0.0

    limite_inferior_2 = math.floor(
        ((-n / z) - 3) / 4
    )

    limite_superior_2 = math.floor(
        ((n / z) - 1) / 4
    )

    for k in range(
        limite_inferior_2,
        limite_superior_2 + 1
    ):
        termo_superior = norm.cdf(
            ((4 * k + 3) * z) / raiz_n
        )

        termo_inferior = norm.cdf(
            ((4 * k + 1) * z) / raiz_n
        )

        soma_2 += (
            termo_superior
            - termo_inferior
        )

    p_valor = 1 - soma_1 + soma_2

    # Evita pequenos erros numéricos fora do intervalo
    p_valor = max(0.0, min(1.0, p_valor))

    return p_valor


# TESTE NA DIREÇÃO DIRETA

def executar_teste_direto(sequencia):
    _, somas = criar_somas_acumuladas(sequencia)

    maior_excursao = int(
        np.max(np.abs(somas))
    )

    p_valor = calcular_p_valor_somas_acumuladas(
        len(sequencia),
        maior_excursao
    )

    return {
        "somas": somas,
        "maior_excursao": maior_excursao,
        "soma_final": int(somas[-1]),
        "p_valor": p_valor
    }


# TESTE NA DIREÇÃO REVERSA

def executar_teste_reverso(sequencia):
    sequencia_reversa = sequencia[::-1]

    _, somas = criar_somas_acumuladas(
        sequencia_reversa
    )

    maior_excursao = int(
        np.max(np.abs(somas))
    )

    p_valor = calcular_p_valor_somas_acumuladas(
        len(sequencia_reversa),
        maior_excursao
    )

    return {
        "somas": somas,
        "maior_excursao": maior_excursao,
        "soma_final": int(somas[-1]),
        "p_valor": p_valor
    }


# EXECUÇÃO DO ALGORITMO 13

resultado_direto = executar_teste_direto(bits)
resultado_reverso = executar_teste_reverso(bits)


# RESULTADOS

print("TESTE DE SOMAS ACUMULADAS")
print("CUMULATIVE SUMS TEST — PRNG")

print(
    f"\nQuantidade de bits: "
    f"{len(bits)}"
)

print("\nDIREÇÃO DIRETA")

print(
    f"Soma acumulada final: "
    f"{resultado_direto['soma_final']}"
)

print(
    f"Maior excursão absoluta: "
    f"{resultado_direto['maior_excursao']}"
)

print(
    f"P-valor: "
    f"{resultado_direto['p_valor']:.10f}"
)

print("\nDIREÇÃO REVERSA")

print(
    f"Soma acumulada final: "
    f"{resultado_reverso['soma_final']}"
)

print(
    f"Maior excursão absoluta: "
    f"{resultado_reverso['maior_excursao']}"
)

print(
    f"P-valor: "
    f"{resultado_reverso['p_valor']:.10f}"
)


# INTERPRETAÇÃO

aprovado_direto = (
    resultado_direto["p_valor"] >= ALFA
)

aprovado_reverso = (
    resultado_reverso["p_valor"] >= ALFA
)

print("\nINTERPRETAÇÃO")

if aprovado_direto:
    print(
        "Direção direta: sequência aprovada."
    )
else:
    print(
        "Direção direta: sequência reprovada."
    )

if aprovado_reverso:
    print(
        "Direção reversa: sequência aprovada."
    )
else:
    print(
        "Direção reversa: sequência reprovada."
    )

if aprovado_direto and aprovado_reverso:
    print(
        "\nResultado geral: sequência aprovada "
        "no teste de somas acumuladas."
    )
else:
    print(
        "\nResultado geral: sequência reprovada "
        "no teste de somas acumuladas."
    )


# GRÁFICO 1 — CAMINHADA ACUMULADA DIRETA

plt.figure(figsize=(13, 6))

plt.plot(
    range(1, len(bits) + 1),
    resultado_direto["somas"],
    linewidth=0.7,
    label="Soma acumulada direta"
)

plt.axhline(
    y=0,
    linestyle="--",
    label="Ponto de equilíbrio"
)

plt.axhline(
    y=resultado_direto["maior_excursao"],
    linestyle=":",
    label="Maior excursão positiva"
)

plt.axhline(
    y=-resultado_direto["maior_excursao"],
    linestyle=":",
    label="Maior excursão negativa"
)

plt.title(
    "Somas Acumuladas\n"
    "Direção direta"
)

plt.xlabel("Posição do bit")
plt.ylabel("Soma acumulada")

plt.grid(alpha=0.3)
plt.legend()

plt.tight_layout()
plt.show()


# GRÁFICO 2 — CAMINHADA ACUMULADA REVERSA

plt.figure(figsize=(13, 6))

plt.plot(
    range(1, len(bits) + 1),
    resultado_reverso["somas"],
    linewidth=0.7,
    label="Soma acumulada reversa"
)

plt.axhline(
    y=0,
    linestyle="--",
    label="Ponto de equilíbrio"
)

plt.axhline(
    y=resultado_reverso["maior_excursao"],
    linestyle=":",
    label="Maior excursão positiva"
)

plt.axhline(
    y=-resultado_reverso["maior_excursao"],
    linestyle=":",
    label="Maior excursão negativa"
)

plt.title(
    "Somas Acumuladas\n"
    "Direção reversa"
)

plt.xlabel("Posição do bit na sequência reversa")
plt.ylabel("Soma acumulada")

plt.grid(alpha=0.3)
plt.legend()

plt.tight_layout()
plt.show()


# GRÁFICO 3 — COMPARAÇÃO DOS P-VALORES

categorias = [
    "Direção direta",
    "Direção reversa"
]

p_valores = [
    resultado_direto["p_valor"],
    resultado_reverso["p_valor"]
]

plt.figure(figsize=(8, 5))

barras = plt.bar(
    categorias,
    p_valores
)

plt.axhline(
    y=ALFA,
    linestyle="--",
    label=f"Limite de aprovação: {ALFA}"
)

plt.title(
    "P-valores do Teste "
    "de Somas Acumuladas"
)

plt.ylabel("P-valor")
plt.ylim(0, 1)

for barra, valor in zip(
    barras,
    p_valores
):
    plt.text(
        barra.get_x()
        + barra.get_width() / 2,
        valor + 0.02,
        f"{valor:.6f}",
        ha="center"
    )

plt.legend()
plt.tight_layout()
plt.show()
```

```python
#  TESTE ESPECTRAL COM TRANSFORMADA DE FOURIER
# DISCRETE FOURIER TRANSFORM TEST — PRNG

import random
import math
import numpy as np
import matplotlib.pyplot as plt
from scipy.special import erfc


# Configurações

QUANTIDADE_BITS = 1_000_000
SEMENTE = 42
ALFA = 0.05


# Geração dos bits pseudoaleatórios

random.seed(SEMENTE)

bits = [
    random.randint(0, 1)
    for _ in range(QUANTIDADE_BITS)
]


# Validação

if len(bits) == 0:
    raise ValueError("A sequência está vazia.")

if any(bit not in (0, 1) for bit in bits):
    raise ValueError(
        "A sequência deve conter apenas bits 0 e 1."
    )


# Conversão dos bits

# O teste transforma:
# 0 -> -1
# 1 -> +1

sequencia_numerica = np.array(
    [
        1 if bit == 1 else -1
        for bit in bits
    ],
    dtype=float
)


# Transformada rápida de Fourier

transformada = np.fft.fft(
    sequencia_numerica
)


# Utiliza apenas a primeira metade do espectro

metade = len(bits) // 2

magnitudes = np.abs(
    transformada[:metade]
)


# Limite espectral

# Aproximadamente 95% das magnitudes devem ficar
# abaixo desse limite em uma sequência aleatória.

limite = math.sqrt(
    math.log(1 / 0.05)
    * len(bits)
)


# Número esperado de magnitudes abaixo do limite

quantidade_esperada = (
    0.95
    * len(bits)
    / 2
)


# Número observado abaixo do limite

quantidade_observada = int(
    np.sum(magnitudes < limite)
)


# Diferença normalizada

diferenca_normalizada = (
    quantidade_observada
    - quantidade_esperada
) / math.sqrt(
    len(bits)
    * 0.95
    * 0.05
    / 4
)


# P-valor

p_valor = erfc(
    abs(diferenca_normalizada)
    / math.sqrt(2)
)


# Quantidade de picos acima do limite

picos_acima_limite = int(
    np.sum(magnitudes >= limite)
)

picos_esperados = (
    0.05
    * len(bits)
    / 2
)


# Resultados

print(
    " TESTE ESPECTRAL "
    "COM TRANSFORMADA DE FOURIER"
)

print(
    "\nQuantidade de bits:",
    len(bits)
)

print(
    "Quantidade de frequências analisadas:",
    len(magnitudes)
)

print(
    f"\nLimite espectral: "
    f"{limite:.6f}"
)

print(
    f"Magnitudes abaixo do limite: "
    f"{quantidade_observada}"
)

print(
    f"Quantidade esperada abaixo do limite: "
    f"{quantidade_esperada:.2f}"
)

print(
    f"\nPicos acima do limite: "
    f"{picos_acima_limite}"
)

print(
    f"Picos esperados acima do limite: "
    f"{picos_esperados:.2f}"
)

print(
    f"\nDiferença normalizada: "
    f"{diferenca_normalizada:.6f}"
)

print(
    f"P-valor: "
    f"{p_valor:.10f}"
)


# Interpretação

if p_valor >= ALFA:
    print(
        "\nResultado: sequência aprovada "
        "no teste espectral."
    )
else:
    print(
        "\nResultado: sequência reprovada "
        "no teste espectral."
    )


# Identificação das maiores frequências

frequencias = np.fft.fftfreq(
    len(bits)
)[:metade]

quantidade_maiores_picos = 10

indices_maiores_picos = np.argsort(
    magnitudes
)[-quantidade_maiores_picos:][::-1]


print("\nDez maiores picos espectrais:")

for posicao, indice in enumerate(
    indices_maiores_picos,
    start=1
):
    print(
        f"{posicao:2d}. "
        f"Frequência = {frequencias[indice]:.8f} | "
        f"Magnitude = {magnitudes[indice]:.6f}"
    )


# Gráfico do espectro completo

plt.figure(
    figsize=(13, 6)
)

plt.plot(
    frequencias,
    magnitudes,
    linewidth=0.5,
    label="Magnitude da FFT"
)

plt.axhline(
    y=limite,
    linestyle="--",
    label="Limite de 95%"
)

plt.title(
    " Espectro da Sequência Pseudoaleatória"
)

plt.xlabel(
    "Frequência normalizada"
)

plt.ylabel(
    "Magnitude"
)

plt.grid(
    alpha=0.3
)

plt.legend()

plt.tight_layout()

plt.show()


# Gráfico das primeiras frequências

quantidade_exibida = 5000

plt.figure(
    figsize=(13, 6)
)

plt.plot(
    frequencias[:quantidade_exibida],
    magnitudes[:quantidade_exibida],
    linewidth=0.8,
    label="Magnitude da FFT"
)

plt.axhline(
    y=limite,
    linestyle="--",
    label="Limite de 95%"
)

plt.title(
    " Primeiras Frequências do Espectro"
)

plt.xlabel(
    "Frequência normalizada"
)

plt.ylabel(
    "Magnitude"
)

plt.grid(
    alpha=0.3
)

plt.legend()

plt.tight_layout()

plt.show()


# Gráfico observado versus esperado

categorias = [
    "Abaixo do limite",
    "Acima do limite"
]

valores_observados = [
    quantidade_observada,
    picos_acima_limite
]

valores_esperados = [
    quantidade_esperada,
    picos_esperados
]

posicoes = np.arange(
    len(categorias)
)

largura = 0.35

plt.figure(
    figsize=(9, 5)
)

plt.bar(
    posicoes - largura / 2,
    valores_observados,
    width=largura,
    label="Observado"
)

plt.bar(
    posicoes + largura / 2,
    valores_esperados,
    width=largura,
    label="Esperado"
)

plt.xticks(
    posicoes,
    categorias
)

plt.title(
    " Frequências Observadas e Esperadas"
)

plt.ylabel(
    "Quantidade de magnitudes"
)

plt.legend()

plt.tight_layout()

plt.show()


# Gráfico do p-valor

plt.figure(
    figsize=(7, 5)
)

barra = plt.bar(
    ["P-valor"],
    [p_valor]
)

plt.axhline(
    y=ALFA,
    linestyle="--",
    label=f"Limite de aprovação: {ALFA}"
)

plt.title(
    " Resultado do Teste Espectral"
)

plt.ylabel(
    "P-valor"
)

plt.ylim(
    0,
    1
)

plt.text(
    barra[0].get_x()
    + barra[0].get_width() / 2,
    p_valor + 0.02,
    f"{p_valor:.6f}",
    ha="center"
)

plt.legend()

plt.tight_layout()

plt.show()
```

```python
# TESTE DE POSTO DE MATRIZES BINÁRIAS

import random
import math
import numpy as np
import matplotlib.pyplot as plt
from scipy.special import gammaincc


# Configurações

QUANTIDADE_BITS = 1_000_000
SEMENTE = 42
ALFA = 0.05

LINHAS = 32
COLUNAS = 32


# Geração dos bits pseudoaleatórios

random.seed(SEMENTE)

bits = [
    random.randint(0, 1)
    for _ in range(QUANTIDADE_BITS)
]


# Função para calcular o posto de uma matriz em GF(2)

def calcular_posto_gf2(matriz):

    matriz = matriz.copy().astype(np.uint8)

    linhas, colunas = matriz.shape

    posto = 0
    coluna_atual = 0

    while (
        posto < linhas
        and coluna_atual < colunas
    ):

        pivo = None

        for linha in range(
            posto,
            linhas
        ):
            if matriz[
                linha,
                coluna_atual
            ] == 1:

                pivo = linha
                break

        if pivo is None:

            coluna_atual += 1
            continue

        if pivo != posto:

            matriz[
                [posto, pivo]
            ] = matriz[
                [pivo, posto]
            ]

        for linha in range(linhas):

            if (
                linha != posto
                and matriz[
                    linha,
                    coluna_atual
                ] == 1
            ):

                matriz[linha] ^= matriz[posto]

        posto += 1
        coluna_atual += 1

    return posto


# Tamanho de cada matriz

BITS_POR_MATRIZ = (
    LINHAS
    * COLUNAS
)


# Quantidade de matrizes completas

quantidade_matrizes = (
    len(bits)
    // BITS_POR_MATRIZ
)


# Bits utilizados

bits_utilizados = bits[
    :quantidade_matrizes
    * BITS_POR_MATRIZ
]


# Cálculo do posto de cada matriz

postos = []

for indice in range(
    quantidade_matrizes
):

    inicio = (
        indice
        * BITS_POR_MATRIZ
    )

    fim = (
        inicio
        + BITS_POR_MATRIZ
    )

    bloco = bits_utilizados[
        inicio:fim
    ]

    matriz = np.array(
        bloco,
        dtype=np.uint8
    ).reshape(
        LINHAS,
        COLUNAS
    )

    posto = calcular_posto_gf2(
        matriz
    )

    postos.append(
        posto
    )


# Categorias utilizadas pelo teste NIST

posto_completo = sum(
    posto == 32
    for posto in postos
)

posto_31 = sum(
    posto == 31
    for posto in postos
)

posto_menor = sum(
    posto <= 30
    for posto in postos
)


frequencias_observadas = [
    posto_completo,
    posto_31,
    posto_menor
]


# Probabilidades esperadas para matrizes 32x32

probabilidades_esperadas = [
    0.2887880950866024,
    0.5775761901732048,
    0.1336357147401928
]


frequencias_esperadas = [
    quantidade_matrizes
    * probabilidade
    for probabilidade
    in probabilidades_esperadas
]


# Estatística qui-quadrado

qui_quadrado = sum(

    (
        observado
        - esperado
    ) ** 2
    / esperado

    for observado, esperado
    in zip(
        frequencias_observadas,
        frequencias_esperadas
    )
)


# Existem três categorias

graus_liberdade = 2


# P-valor

p_valor = gammaincc(
    graus_liberdade / 2,
    qui_quadrado / 2
)


# Resultados

print(
    "TESTE DE POSTO DE MATRIZES BINÁRIAS"
)

print(
    "\nQuantidade total de bits:",
    len(bits)
)

print(
    "Bits por matriz:",
    BITS_POR_MATRIZ
)

print(
    "Dimensão das matrizes:",
    f"{LINHAS} x {COLUNAS}"
)

print(
    "Quantidade de matrizes:",
    quantidade_matrizes
)

print(
    "Bits utilizados:",
    len(bits_utilizados)
)

print(
    "Bits descartados:",
    len(bits)
    - len(bits_utilizados)
)


print(
    "\nPosto 32:"
)

print(
    "Observado:",
    posto_completo
)

print(
    "Esperado:",
    f"{frequencias_esperadas[0]:.2f}"
)


print(
    "\nPosto 31:"
)

print(
    "Observado:",
    posto_31
)

print(
    "Esperado:",
    f"{frequencias_esperadas[1]:.2f}"
)


print(
    "\nPosto <= 30:"
)

print(
    "Observado:",
    posto_menor
)

print(
    "Esperado:",
    f"{frequencias_esperadas[2]:.2f}"
)


print(
    f"\nQui-quadrado: "
    f"{qui_quadrado:.6f}"
)

print(
    f"P-valor: "
    f"{p_valor:.10f}"
)


# Interpretação

if p_valor >= ALFA:

    print(
        "\nResultado: sequência aprovada "
        "no teste de posto de matrizes binárias."
    )

else:

    print(
        "\nResultado: sequência reprovada "
        "no teste de posto de matrizes binárias."
    )


# Gráfico das frequências observadas e esperadas

categorias = [
    "Posto 32",
    "Posto 31",
    "Posto <= 30"
]

posicoes = np.arange(
    len(categorias)
)

largura = 0.35


plt.figure(
    figsize=(9, 5)
)

plt.bar(
    posicoes - largura / 2,
    frequencias_observadas,
    largura,
    label="Observado"
)

plt.bar(
    posicoes + largura / 2,
    frequencias_esperadas,
    largura,
    label="Esperado"
)

plt.xticks(
    posicoes,
    categorias
)

plt.title(
    "Posto das Matrizes Binárias"
)

plt.ylabel(
    "Quantidade de matrizes"
)

plt.legend()

plt.grid(
    axis="y",
    alpha=0.3
)

plt.tight_layout()

plt.show()


# Distribuição completa dos postos encontrados

postos_unicos = sorted(
    set(postos)
)

quantidades = [
    postos.count(posto)
    for posto in postos_unicos
]


plt.figure(
    figsize=(10, 5)
)

plt.bar(
    postos_unicos,
    quantidades
)

plt.title(
    "Distribuição dos Postos das Matrizes"
)

plt.xlabel(
    "Posto da matriz"
)

plt.ylabel(
    "Quantidade"
)

plt.xticks(
    postos_unicos
)

plt.grid(
    axis="y",
    alpha=0.3
)

plt.tight_layout()

plt.show()


# Gráfico do p-valor

plt.figure(
    figsize=(7, 5)
)

barra = plt.bar(
    ["P-valor"],
    [p_valor]
)

plt.axhline(
    y=ALFA,
    linestyle="--",
    label="Limite de aprovação: 0,05"
)

plt.title(
    "Resultado do Binary Matrix Rank Test"
)

plt.ylabel(
    "P-valor"
)

plt.ylim(
    0,
    1
)

plt.text(
    barra[0].get_x()
    + barra[0].get_width() / 2,
    p_valor + 0.02,
    f"{p_valor:.6f}",
    ha="center"
)

plt.legend()

plt.tight_layout()

plt.show()
```

```python
# TESTE DE PADRÕES NÃO SOBREPOSTOS
# NON-OVERLAPPING TEMPLATE MATCHING TEST — PRNG

import random
import math
import numpy as np
import matplotlib.pyplot as plt
from scipy.special import gammaincc


# Configurações

QUANTIDADE_BITS = 1_000_000
SEMENTE = 42
ALFA = 0.05

TAMANHO_BLOCO = 10_000

# Padrão que será procurado
PADRAO = [0, 0, 0, 0, 0, 0, 0, 0, 1]

TAMANHO_PADRAO = len(PADRAO)


# Geração dos bits pseudoaleatórios

random.seed(SEMENTE)

bits = [
    random.randint(0, 1)
    for _ in range(QUANTIDADE_BITS)
]


# Quantidade de blocos

quantidade_blocos = (
    len(bits)
    // TAMANHO_BLOCO
)


bits_utilizados = bits[
    :quantidade_blocos
    * TAMANHO_BLOCO
]


# Função para contar padrões não sobrepostos

def contar_padrao_nao_sobreposto(
    bloco,
    padrao
):

    tamanho_padrao = len(
        padrao
    )

    contador = 0
    indice = 0

    while (
        indice
        <= len(bloco)
        - tamanho_padrao
    ):

        trecho = bloco[
            indice:
            indice + tamanho_padrao
        ]

        if trecho == padrao:

            contador += 1

            # Pula o padrão inteiro para impedir
            # sobreposição
            indice += tamanho_padrao

        else:

            indice += 1

    return contador


# Contagem do padrão em cada bloco

contagens = []

for numero_bloco in range(
    quantidade_blocos
):

    inicio = (
        numero_bloco
        * TAMANHO_BLOCO
    )

    fim = (
        inicio
        + TAMANHO_BLOCO
    )

    bloco = bits_utilizados[
        inicio:fim
    ]

    quantidade = (
        contar_padrao_nao_sobreposto(
            bloco,
            PADRAO
        )
    )

    contagens.append(
        quantidade
    )


# Média esperada

media_esperada = (
    TAMANHO_BLOCO
    - TAMANHO_PADRAO
    + 1
) / (
    2 ** TAMANHO_PADRAO
)


# Variância esperada

variancia_esperada = (
    TAMANHO_BLOCO
    * (
        1 / (2 ** TAMANHO_PADRAO)
        - (
            2 * TAMANHO_PADRAO - 1
        )
        / (
            2 ** (
                2 * TAMANHO_PADRAO
            )
        )
    )
)


# Estatística qui-quadrado

qui_quadrado = sum(

    (
        quantidade
        - media_esperada
    ) ** 2
    / variancia_esperada

    for quantidade
    in contagens
)


# P-valor

p_valor = gammaincc(
    quantidade_blocos / 2,
    qui_quadrado / 2
)


# Resultados

print(
    "TESTE DE PADRÕES NÃO SOBREPOSTOS"
)

print(
    "\nQuantidade total de bits:",
    len(bits)
)

print(
    "Bits utilizados:",
    len(bits_utilizados)
)

print(
    "Tamanho dos blocos:",
    TAMANHO_BLOCO
)

print(
    "Quantidade de blocos:",
    quantidade_blocos
)

print(
    "Padrão analisado:",
    "".join(
        str(bit)
        for bit in PADRAO
    )
)

print(
    "Tamanho do padrão:",
    TAMANHO_PADRAO
)


print(
    f"\nMédia esperada por bloco: "
    f"{media_esperada:.6f}"
)

print(
    f"Variância esperada: "
    f"{variancia_esperada:.6f}"
)


print(
    "\nPrimeiras 20 contagens:"
)

for indice, quantidade in enumerate(
    contagens[:20],
    start=1
):

    print(
        f"Bloco {indice:2d}: "
        f"{quantidade}"
    )


print(
    f"\nQui-quadrado: "
    f"{qui_quadrado:.6f}"
)

print(
    f"P-valor: "
    f"{p_valor:.10f}"
)


# Interpretação

if p_valor >= ALFA:

    print(
        "\nResultado: sequência aprovada "
        "no teste de padrões não sobrepostos."
    )

else:

    print(
        "\nResultado: sequência reprovada "
        "no teste de padrões não sobrepostos."
    )


# Gráfico das ocorrências por bloco

indices_blocos = list(
    range(
        1,
        quantidade_blocos + 1
    )
)


plt.figure(
    figsize=(13, 6)
)

plt.plot(
    indices_blocos,
    contagens,
    marker="o",
    markersize=3,
    linewidth=0.8,
    label="Ocorrências observadas"
)

plt.axhline(
    y=media_esperada,
    linestyle="--",
    label="Média esperada"
)

plt.title(
    "Teste de Padrões Não Sobrepostos — PRNG"
)

plt.xlabel(
    "Bloco"
)

plt.ylabel(
    "Quantidade de ocorrências"
)

plt.grid(
    alpha=0.3
)

plt.legend()

plt.tight_layout()

plt.show()


# Histograma das ocorrências

plt.figure(
    figsize=(10, 5)
)

plt.hist(
    contagens,
    bins=20,
    edgecolor="black"
)

plt.axvline(
    x=media_esperada,
    linestyle="--",
    label="Média esperada"
)

plt.title(
    "Distribuição das Ocorrências do Padrão"
)

plt.xlabel(
    "Ocorrências por bloco"
)

plt.ylabel(
    "Quantidade de blocos"
)

plt.legend()

plt.grid(
    axis="y",
    alpha=0.3
)

plt.tight_layout()

plt.show()


# Comparação da média observada com a esperada

media_observada = np.mean(
    contagens
)

categorias = [
    "Média observada",
    "Média esperada"
]

valores = [
    media_observada,
    media_esperada
]


plt.figure(
    figsize=(8, 5)
)

barras = plt.bar(
    categorias,
    valores
)

plt.title(
    "Frequência Média do Padrão"
)

plt.ylabel(
    "Ocorrências por bloco"
)

for barra, valor in zip(
    barras,
    valores
):

    plt.text(
        barra.get_x()
        + barra.get_width() / 2,
        valor,
        f"{valor:.4f}",
        ha="center",
        va="bottom"
    )

plt.tight_layout()

plt.show()


# Gráfico do p-valor

plt.figure(
    figsize=(7, 5)
)

barra = plt.bar(
    ["P-valor"],
    [p_valor]
)

plt.axhline(
    y=ALFA,
    linestyle="--",
    label="Limite de aprovação: 0,05"
)

plt.title(
    "Resultado do Teste de Padrões Não Sobrepostos"
)

plt.ylabel(
    "P-valor"
)

plt.ylim(
    0,
    1
)

plt.text(
    barra[0].get_x()
    + barra[0].get_width() / 2,
    p_valor + 0.02,
    f"{p_valor:.6f}",
    ha="center"
)

plt.legend()

plt.tight_layout()

plt.show()
```

# Experimento padronizado

30 lotes × 10.000 bits. A sequência é gerada uma única vez por lote e todas as métricas são calculadas sobre o mesmo lote.


```python

import math
import os
from pathlib import Path

import numpy as np
import pandas as pd
import statsmodels
import scipy


NUM_LOTES = 30
BITS_POR_LOTE = 10_000
ALFA = 0.05
Z_95 = 1.959963984540054

print("Python/ambiente estatístico:")
print("statsmodels:", statsmodels.__version__)
print("scipy:", scipy.__version__)


def intervalo_wilson(p, n, z=Z_95):
    if n == 0:
        return np.nan, np.nan

    denominador = 1 + (z ** 2) / n
    centro = (p + (z ** 2) / (2 * n)) / denominador
    margem = (
        z
        * math.sqrt(
            (p * (1 - p) + (z ** 2) / (4 * n))
            / n
        )
        / denominador
    )

    return centro - margem, centro + margem


def runs_test(bits):
    """
    Runs Test (Wald-Wolfowitz) usando statsmodels.

    H0: a ordem dos bits é compatível com aleatoriedade em relação
        ao número de corridas.
    H1: o número de corridas difere do esperado sob H0.

    Regra de decisão:
        p >= ALFA -> não rejeita H0
        p <  ALFA -> rejeita H0
    """
    from statsmodels.sandbox.stats.runs import runstest_1samp

    vetor = np.fromiter(
        (1 if bit == "1" else 0 for bit in bits),
        dtype=np.int8
    )

    if len(vetor) < 2 or np.all(vetor == vetor[0]):
        return np.nan, np.nan, np.nan

    numero_runs = 1 + int(np.sum(vetor[1:] != vetor[:-1]))

    z, p_valor = runstest_1samp(
        vetor,
        cutoff=0.5,
        correction=False
    )

    return numero_runs, float(z), float(p_valor)



def calcular_metricas(bits, fonte, lote, condicao="bruta"):
    bits = "".join(bits)

    n = len(bits)

    if n == 0:
        raise ValueError("O lote não contém bits.")

    zeros = bits.count("0")
    uns = bits.count("1")

    p0 = zeros / n
    p1 = uns / n

    vies = abs(p1 - 0.5)

    shannon = 0.0

    for p in (p0, p1):
        if p > 0:
            shannon -= p * math.log2(p)

    min_entropia = -math.log2(
        max(p0, p1)
    )

    vetor = np.fromiter(
        (1 if bit == "1" else 0 for bit in bits),
        dtype=np.float64
    )

    if (
        n > 1
        and np.std(vetor[:-1]) > 0
        and np.std(vetor[1:]) > 0
    ):
        autocorrelacao_lag1 = float(
            np.corrcoef(
                vetor[:-1],
                vetor[1:]
            )[0, 1]
        )
    else:
        autocorrelacao_lag1 = np.nan

    numero_runs, runs_z, runs_p_valor = runs_test(bits)

    ic95_inferior, ic95_superior = intervalo_wilson(
        p1,
        n
    )

    return {
        "fonte": fonte,
        "condicao": condicao,
        "lote": lote,
        "n_bits": n,
        "zeros": zeros,
        "uns": uns,
        "p0": p0,
        "p1": p1,
        "vies": vies,
        "ic95_p1_inferior": ic95_inferior,
        "ic95_p1_superior": ic95_superior,
        "entropia_shannon": shannon,
        "min_entropia_empirica": min_entropia,
        "autocorrelacao_lag1": autocorrelacao_lag1,
        "numero_runs": numero_runs,
        "runs_z": runs_z,
        "runs_p_valor": runs_p_valor,
        "runs_aprovado_alfa_005": (
            runs_p_valor >= ALFA
            if not np.isnan(runs_p_valor)
            else False
        ),
    }


def salvar_lote(bits, pasta, nome):
    pasta = Path(pasta)
    pasta.mkdir(
        parents=True,
        exist_ok=True
    )

    caminho = pasta / nome

    caminho.write_text(
        "".join(bits),
        encoding="utf-8"
    )

    return caminho


def resumo_metricas(df):
    colunas = [
        "p1",
        "vies",
        "entropia_shannon",
        "min_entropia_empirica",
        "autocorrelacao_lag1",
        "runs_z",
        "runs_p_valor",
    ]

    print("\nResumo dos 30 lotes:")
    display(
        df[colunas]
        .agg(["mean", "std", "min", "median", "max"])
        .T
    )

    print(
        "\nProporção de aprovação no Runs Test (alfa = 0,05):",
        f"{df['runs_aprovado_alfa_005'].mean():.2%}"
    )


# EXPERIMENTO PADRONIZADO — PRNG DIDÁTICO

import random

SEMENTE = 7
FONTE = "PRNG_Mersenne_Twister_seed_7"

pasta_saida = Path(
    "/content/dados_padronizados/prng"
)

gerador = random.Random(SEMENTE)

resultados = []

for lote in range(1, NUM_LOTES + 1):

    bits = "".join(
        str(gerador.getrandbits(1))
        for _ in range(BITS_POR_LOTE)
    )

    salvar_lote(
        bits,
        pasta_saida,
        f"prng_lote_{lote:02d}.txt"
    )

    resultados.append(
        calcular_metricas(
            bits,
            fonte=FONTE,
            lote=lote
        )
    )

df_prng = pd.DataFrame(resultados)

caminho_csv = (
    pasta_saida
    / "prng_metricas_30x10000.csv"
)

df_prng.to_csv(
    caminho_csv,
    index=False
)

print("Experimento concluído.")
print("Semente:", SEMENTE)
print("Lotes:", NUM_LOTES)
print("Bits por lote:", BITS_POR_LOTE)
print("Total de bits:", NUM_LOTES * BITS_POR_LOTE)
print("CSV:", caminho_csv)

display(df_prng.head())

resumo_metricas(df_prng)
```
